```python
'''
 Task 4 - Filtering with Conditions
13. Find students with Final_Marks at least 70.
14. Find students with Attendance_pct below 70 AND Study_Hours above 8.
15. Find students from Lahore OR Islamabad after accounting for inconsistent text values.
16. Use isin() to select two or more courses.
17. Use between() for a sensible numeric range of your choice.
18. Filter students who have Internet_Access = No or a missing Internet_Access value.
Challenge: Create one compound filter containing at least three conditions using &, |, and ~ correctly.
'''

df.loc[df["Final_Marks"] >= 70].head()

df.loc[(df["Attendance_pct"] < 70) & (df['Study_Hours']>8)]

df.loc[(df['City']== 'Lahore') | (df['City']== 'Islamabad'), ['Name', 'City']]

df.loc[df['Course'].isin(['Data Science', 'Web Dev']), ['Name','Course']] # it act like multiple logical OR operation

# so i select the age column
df.loc[df['Age'].between(25,30), ['Name','Age']]

#filtering student who have no internet access

df.loc[(df['Internet_Access']== 'No') | (df['Internet_Access'].isna()), ['Name', 'Internet_Access']] #using isna or is null return same output

# challenge creating compound filter # studnet who have assingment score greater than 70 quiz score greater than 70 or has internate access and not in the both of the course
df.loc[(df['Assignment_Score'] >= 70) & (df['Quiz_Score'] >= 70) | (df['Internet_Access'] == 'Yes') & (~df['Course'].isin(['Data Science', 'Web Dev']))].head()
```
