```python
'''
Task 10 - Validate Ranges and Handle Invalid Values
42. Define sensible valid ranges for Attendance_pct, Assignment_Score, Quiz_Score, Previous_Marks,
Final_Marks, Age, and Satisfaction_Rating.
43. Find rows that violate those rules.
44. Do not silently clip everything. Decide whether each invalid value should be corrected, set missing, or
investigated.
45. After your decision, verify the cleaned ranges with min/max or describe().
Challenge: Write a reusable boolean validity check for at least two columns.
Skill focus: domain validation, outliers vs invalid values, careful cleaning
'''

#42 valid range i.e 0 to 100
valid_ranges = {
    'Age':(15,50),
    'Attendance_pct': (0,100),
    'Quiz_Score': (0,100),
    'Previous_Marks': (0,100),
    'Satisfaction_Rating': (1,5),
    'Final_Marks': (0,100),
}
#chellange
def check_validity (df, column, minval, maxval):
    if column not in df.columns:
        raise ValueError("column not exist")
    return df[column].between(minval, maxval)

#43 find all rows that voilate these rules
invalid_mask = pd.Series(False, index = df.index)

for col, (minv,maxv) in valid_ranges.items():
    if col in df.columns:
        col_invalid = ~check_validity(df,col,minv,maxv)
        invalid_mask = invalid_mask | col_invalid
        numofviolations = col_invalid.sum()
        if numofviolations > 0:
            print(f'column : {col} has violations: {numofviolations} invalid values')


invalid_rows = df[invalid_mask]
print(f'total no of invalid rows{len(invalid_rows)}')
```

```python
#44 strategic clean
import numpy as np
df_clean = df.copy()

for col, (minv,maxv) in valid_ranges.items():
    if col not in df_clean.columns:
        continue
    
    #identify exactly violation rows for respected cols
    invalid_idx = df_clean[~check_validity(df_clean, col, minv,maxv)].index
    for idx in invalid_idx:
        val = df_clean.loc[idx, col]

        # a strategy for satisfaction rating
        if col == 'Satisfaction_Rating' and val > 5 and val <=50:
            df_clean.loc[idx, col] = val / 10.0
        elif col in ['Attendance_pct', 'Assignment_Score', 'Quiz_Score'] and (maxv < val <= maxv+5):
            df_clean.loc[idx,col] = maxv
        else:
            df_clean.loc[idx,col] = np.nan

#45 verification summary

df_clean[list[valid_ranges.keys()]].describe()
```
